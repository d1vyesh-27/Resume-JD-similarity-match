## 1. Phase 1: Preprocessing Experiments & Validation
Test controlled preprocessing, the alias dictionary, and optional lemmatization.

```python
# 1. Define a tricky test string that has symbols and punctuation
test_text = """
Candidate has 5 years of experience in software development. 
Proficient in Python, C++, C#, and .NET. 
Familiar with Node.js, scikit-learn, and REST APIs. 
Also worked with PLC/SCADA systems in an ISO 9001 environment.
"""

# 2. Define our Alias Dictionary to protect technical terms from being destroyed
TECHNICAL_ALIASES = {
    "c++": "cplusplus",
    "c#": "csharp",
    ".net": "dotnet",
    "node.js": "nodejs",
    "scikit-learn": "scikitlearn",
    "rest apis": "rest api",       # Normalize plural
    "restful api": "rest api",     # Normalize equivalent
    "machine-learning": "machine learning"
}

print("Original text: ")
print(test_text)
```

```python
# 1. Lowercase the text
cleaned_text = test_text.lower()

# 2. Map the technical aliases
for original_term, safe_term in TECHNICAL_ALIASES.items():
    cleaned_text = cleaned_text.replace(original_term,safe_term)
    # We replace the original term with the safe term
    

print("After Lowercasing and Alias Replacement:")
print(cleaned_text)
```

```python
import spacy

# loads small pretrained statistical model of english lang.
# spacy analyzes grammar of sentence, every word becomes a token object that knows properties about itself
nlp = spacy.load("en_core_web_sm")

# splits text into words, tags every single word as noun/verb/etc
# figures how words relate to each other, lemmatization
doc = nlp(cleaned_text)

final_tokens = []
for token in doc:
    if not token.is_punct and not token.is_stop and not token.is_space:
        final_tokens.append(token.lemma)

final_text = " ".join(final_tokens)

print("Final Preprocessed Text: ")
print(final_text)
```
